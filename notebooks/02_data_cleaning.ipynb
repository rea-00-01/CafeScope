```python
import pandas as pd 
import numpy as np 
from pathlib import Path
```

```python
PROJECT_ROOT = Path.cwd().parent

RAW_DATA_DIR = PROJECT_ROOT / "data" / "raw"
INTERIM_DATA_DIR = PROJECT_ROOT / "data" / "interim"
PROCESSED_DATA_DIR = PROJECT_ROOT / "data" / "processed"

print("Project root:", PROJECT_ROOT)
print("Raw data:", RAW_DATA_DIR)
print("Interim data:", INTERIM_DATA_DIR)
print("Processed data:", PROCESSED_DATA_DIR)
```

```python
input_path = INTERIM_DATA_DIR / "cafe_db.csv"
cafe_db = pd.read_csv(input_path, encoding="utf-8")
cafe_db.head(5)
```

```python

cafe_db.dtypes
```

```python
cafe_db.info()
```

```python
cafe_db.columns = (
    cafe_db.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)
```

```python
cafe_db.columns.tolist()
```

```python
text_columns = [
    "name",
    "location",
    "category",
    "cuisine_offered"
]

for col in text_columns:
    cafe_db[col] = (
        cafe_db[col]
        .astype("string")
        .str.strip()
    )
```

```python
cafe_db[text_columns] = cafe_db[text_columns].replace(
    r"^\s*$",
    pd.NA,
    regex=True
)
```

```python
cafe_db["cost_for_two"].value_counts().head(30)
```

```python
numeric_columns = [
    "cost_for_two",
    "avg_din_rate",
    "avg_del_rate"
]

cafe_db[numeric_columns].dtypes
```

```python
for col in numeric_columns:
    cafe_db[col] = pd.to_numeric(
        cafe_db[col],
        errors="coerce"
    )
```

```python
cafe_db[numeric_columns].describe()
```

```python
before = len(cafe_db)

cafe_db = cafe_db.drop_duplicates().copy()

after = len(cafe_db)

print("Rows before:", before)
print("Rows after:", after)
print("Exact duplicates removed:", before - after)
```

```python
missing_summary = pd.DataFrame({
    "missing_count": cafe_db.isna().sum(),
    "missing_percent": (
        cafe_db.isna().mean() * 100
    ).round(2)
}).sort_values(
    "missing_count",
    ascending=False
)

missing_summary
```

```python
print("Final shape:", cafe_db.shape)

print("\nData types:")
print(cafe_db.dtypes)

print("\nExact duplicates:")
print(cafe_db.duplicated().sum())
```

```python
cafe_db.head(10)
```

```python
flag_columns = [
    "is_vegeterian",
    "is_bar_available",
    "home_delivery",
    "offers_dining"
]

for col in flag_columns:
    print(f"\n--- {col} ---")
    print(cafe_db[col].value_counts(dropna=False))
```

```python
flag_columns = [
    "is_vegeterian",
    "is_bar_available",
    "home_delivery",
    "offers_dining"
]

for col in flag_columns:
    cafe_db[col] = (
        cafe_db[col]
        .str.strip()
        .str.lower()
        .map({
            "yes": True,
            "no": False
        })
    )
```

```python
for col in flag_columns:
    print(cafe_db[col].value_counts(dropna=False))
```

```python
for col in ["din_rev", "del_rev"]:
    print(f"\n--- {col} ---")
    print(cafe_db[col].value_counts(dropna=False).head(20))
```

```python
cafe_db["dining_review_count"] = (
    cafe_db["din_rev"]
    .str.extract(r"(\d+)")
    [0]
    .astype("Float64")
)
```

```python
cafe_db[
    ["din_rev", "dining_review_count"]
].head(20)
```

```python
cafe_db["delivery_review_count"] = (
    cafe_db["del_rev"]
    .str.extract(r"([\d,]+)")[0]
    .str.replace(",", "", regex=False)
    .astype("Float64")
)
```

```python
cafe_db[
    ["del_rev", "delivery_review_count"]
].head(20)
```

```python
cafe_db[
    [
        "dining_review_count",
        "delivery_review_count"
    ]
].describe()
```

```python
cafe_db["category"].value_counts(dropna=False)
```

```python
cafe_db["category"] = (
    cafe_db["category"]
    .str.replace("CAF�", "CAFE", regex=False)
    .str.strip()
)
```

```python
cafe_db["category"].value_counts(dropna=False).head(30)
```

```python
print("Cost range:")
print(cafe_db["cost_for_two"].min(), cafe_db["cost_for_two"].max())

print("\nDining rating range:")
print(cafe_db["avg_din_rate"].min(), cafe_db["avg_din_rate"].max())

print("\nDelivery rating range:")
print(cafe_db["avg_del_rate"].min(), cafe_db["avg_del_rate"].max())
```

```python
text_columns = [
    "name",
    "url",
    "location",
    "din_rev",
    "del_rev",
    "category",
    "cuisine_offered"
]

for col in text_columns:
    cafe_db[col] = (
        cafe_db[col]
        .astype("string")
        .str.strip()
    )
```
