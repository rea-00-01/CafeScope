```python
CAFE_TYPE_CUISINES = [
    "Cafe",
    "Beverages",
    "Desserts",
    "Bakery",
    "Coffee",
    "Tea",
]
```

```python
df["is_explicit_cafe"]=(
    df["category"].str.contains("CAF", case=False , na=False)
)
```

```python
df["has_cafe_type_cuisine"] = (
    df["Cuisine_offered"]
    .fillna("")
    .str.contains(
        "|".join(CAFE_TYPE_CUISINES),
        case=False,
        na=False
    )
)
```

```python
df[["is_explicit_cafe", "has_cafe_type_cuisine"]].value_counts()
```

```python
possible_cafes = df[
    (~df["is_explicit_cafe"]) &
    (df["has_cafe_type_cuisine"])
].copy()
```

```python
possible_cuisine_counts = (
    possible_cafes["Cuisine_offered"]
    .fillna("")
    .str.split(",")
    .explode()
    .str.strip()
    .value_counts()
)

possible_cuisine_counts
```

```python
for cuisine in CAFE_TYPE_CUISINES:
    matches = possible_cafes[
        possible_cafes["Cuisine_offered"]
        .fillna("")
        .str.contains(cuisine, case=False, na=False)
    ]

    print(f"\n--- {cuisine}: {len(matches)} ---")
    print(
        matches[
            ["Name", "category", "Cuisine_offered"]
        ].head(10).to_string(index=False)
    )
```

```python
STRONG_NAME_KEYWORDS = [
    "cafe",
    "coffee",
    "coffeehouse",
    "coffee house",
    "chai cafe",
    "tea cafe",
]

MEDIUM_NAME_KEYWORDS = [
    "chai",
    "tea",
]
```

```python
name_pattern = "|".join(STRONG_NAME_KEYWORDS)

df["has_cafe_name_signal"] = (
    df["Name"]
    .fillna("")
    .str.contains(name_pattern, case=False, na=False)
)
```

```python
df[
    [
        "is_explicit_cafe",
        "has_cafe_type_cuisine",
        "has_cafe_name_signal"
    ]
].value_counts()
```

```python
name_only = df[
    (~df["is_explicit_cafe"]) &
    (~df["has_cafe_type_cuisine"]) &
    (df["has_cafe_name_signal"])
].copy()

name_only[["Name", "category", "Cuisine_offered"]].head(30)
```

```python
name_only = df[
    (df["is_explicit_cafe"]) &
    (~df["has_cafe_type_cuisine"]) &
    (~df["has_cafe_name_signal"])
].copy()

name_only[["Name", "category", "Cuisine_offered"]].head(30)
```

```python
name_only = df[
    (df["is_explicit_cafe"]) &
    (df["has_cafe_type_cuisine"]) &
    (~df["has_cafe_name_signal"])
].copy()

name_only[["Name", "category", "Cuisine_offered"]].head(30)
```

```python
# df["is_cafescope_cafe"] = (
#     df["is_explicit_cafe"] |
#     (
#         ~df["is_explicit_cafe"] |
#         df["has_cafe_name_signal"]
#     )
# )

df["is_cafescope_cafe"] = (
    df["is_explicit_cafe"] |
    df["has_cafe_name_signal"]
)

df["is_cafescope_cafe"].value_counts()
```

```python
# Final CafeScope dataset
cafe_df = df[df["is_cafescope_cafe"]].copy()

print("Original dataset:", len(df))
print("CafeScope dataset:", len(cafe_df))
print("Excluded:", len(df) - len(cafe_df))
```
